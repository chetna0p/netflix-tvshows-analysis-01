```python
import pandas as pd

# Load data
df = pd.read_csv("TV Shows - Netflix.csv")

# Clean spaces
df['Titles'] = df['Titles'].str.strip()
df['Rating'] = df['Rating'].str.strip()

# Remove missing values
df = df.dropna()

# Remove duplicates
df = df.drop_duplicates()

# Standardize rating
df['Rating'] = df['Rating'].str.upper()

# Convert Netflix column
df['Netflix'] = df['Netflix'].apply(lambda x: True if x == 1 else False)

# Reset index
df = df.reset_index(drop=True)

# Save cleaned dataset
df.to_csv("TV_Shows_Netflix_Cleaned.csv", index=False)

print("Dataset Cleaned Successfully!")
print(df.describe())
```

```python
#Top Rated Shows
df.sort_values(by="IMDB_Rating", ascending=False).head(10)
#Average IMDB Rating
df['IMDB_Rating'].mean()
#Netflix vs Non-Netflix Shows
df.groupby("Netflix")["IMDB_Rating"].mean()
#Shows Released Each Year
df['Year'].value_counts().sort_index()
```

```python
import matplotlib.pyplot as plt
df['Year'].value_counts().sort_index().plot(kind="line")
plt.title("TV Shows Released Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Shows")
plt.show()
```

```python
df['Rating'].value_counts().plot(kind="bar")
plt.title("Audience Rating Categories")
plt.show()
netflix_shows = df[df['Netflix']==True]
netflix_shows.sort_values(by="IMDB_Rating", ascending=False).head()
```

```python
top10 = df.sort_values(by="IMDB_Rating", ascending=False).head(10)

plt.figure()
plt.barh(top10['Titles'], top10['IMDB_Rating'])
plt.title("Top 10 TV Shows by IMDB Rating")
plt.xlabel("IMDB Rating")
plt.show()
```
